Alternate methods: Convert to SQLite DB:

```python
datafile = 'https://vincentarelbundock.github.io/Rdatasets/csv/HSAUR/toothpaste.csv'

toothpastefile = pd.read_csv(datafile)

with contextlib.closing(db.connect("toothpaste.db")) as db_con:
  toothpastefile.to_sql("toothpasteInfo", db_con, if_exists="replace", index=False)

! ls -l
```

```python
%%script sqlite3 --column --header toothpaste.db
select * from pragma_table_info('toothpasteInfo');
```
