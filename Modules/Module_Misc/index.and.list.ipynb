```python
# @title Setup
import random

DEFAULT_SEED = object()

def index_list( my_list, shuffle=False, random_seed=DEFAULT_SEED ):
  '''Given a list, displays the elements and the indices'''
  local_my_list = my_list.copy()

  if shuffle:
    if random_seed is not DEFAULT_SEED:
      random.seed(random_seed)
    random.shuffle(local_my_list)

  print(f"my_list = {local_my_list}")
  print()
  chosen_item = random.choice(local_my_list)
  print(f"Where is the {chosen_item} in my_list?")
  print("\n"*3)
  print(f"{'Item':<12} | {'Positive Index':<15} | {'Negative Index'}")
  print("-" * 48)

  # enumerate() gives us both the index number and the item at the same time
  length = len(local_my_list)
  for positive_index, item in enumerate(local_my_list):
    # Calculate the corresponding negative index
    negative_index = positive_index - length

    # Print the formatted row side-by-side
    print(f"{item:<12} | {positive_index:<15} | {negative_index}")
  print()
  return local_my_list
```

```python
# A list of items to demonstrate Python indexing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
fruits
```

```python
my_list = index_list(fruits, shuffle=True, random_seed=42 )
```
