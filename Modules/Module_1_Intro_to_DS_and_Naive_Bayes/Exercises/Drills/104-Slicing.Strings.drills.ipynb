## Your turn

```python
for n in outputs.keys():
  s = outputs[n]

  v1 = line.find(s)
  if v1 == -1:
    v1 = len(line) - 1 - line[::-1].find(s)
    v2 = v1 - len(s)
    if v2 == -1:
      v2 = None
    v3 = -1
  else:
    v2 = len(s) + v1
    v3 = 1

  print(f'''
    {n=}
    {s=}
    {line=}
    {v1=}
    {v2=}
    {v3=}
    {line[v1:v2:v3]=}
    {line[v1:v2:v3]==s=}
  ''')
```
