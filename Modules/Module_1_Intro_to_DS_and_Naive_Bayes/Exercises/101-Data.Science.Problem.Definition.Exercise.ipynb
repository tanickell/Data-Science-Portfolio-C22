# Data Science Problem Definition


For each item develop a well-worded data science problem statement.  Your answer should include what type of data science problem this is ( supervised vs unsupervised ) and then whether it is a classification, regression, or clustering problem.  If it is a supervised problem, you should state what the target of the prediction is.

1. Zillow wants you to predict real estate sales prices. They have a large
database of 12 years of home features accompanied by the corresponding
sales prices.


Statement:

Zillow wants to build a system to figure out how much to list a new house for based on past home prices. To do this, they need to make predictions based on previous sales prices relative to home features. This is a supervised regression data science problem whose target is to predict home values.



---

2. Amazon wants you to develop a code that upsells (add additional items to the
customer’s order) based on what items have been viewed and bought by that
customer over the past 2 years.



Statement:

To increase sales, Amazon is implementing a new strategy to suggest additional items based on customer view and purchase history at the time of checkout. This is an unsupervised clustering problem, as it involves determining on-the-fly what items might relate to other items a customer has expressed interest in.



---

3. MGM wants to know what features of a movie seem to increase the movie’s
box office income. They have a large database of previously released movies
accompanied by the resulting box office revenue.



Statement:

MGM is looking to streamline its greenlighting process and wants to make predictions about what types of movies drive the highest ticket sales. To do this, they want to examine attributes of past movies -- genre, length, actors, etc. -- and see how they relate to revenue. This is a supervised regression problem, where the desired prediction is the potential income of a movie based on these features.



---

4. Spotify wants you to develop a method of predicting whether a customer is
about to terminate their service BEFORE they terminate so that they can offer
enticements to the customer to keep the service and not terminate (called
‘churn’)



Statement:

Spotify loses a great number of subscribers to churn every month, and is looking to devise a way to predict the 'mood' of a user to increase retention. This is a supervised classification problem (binary -- are they in a near-termination state or not?), where the goal is to predict this mood. Success would mean the number of churns for a monthly period decreased based on incentives provided to people determined to be in a near-termination mood.



---

5. Comcast wants you to develop a chat bot (automated customer service) to
take text input and respond with the appropriate help and/or service
response.



Statement:

Comcast is looking to broaden its customer service offerings with chat bot services. To do this, it wants to provide automated responses based on user text input. This would be an unsupervised clustering data sciene problem, as it involves taking user data -- here, text iput -- and relating it to other information that is not yet defined. That is, the output can be whatever the model determines may be useful to the user using the chat bot.



---

6. Instagram wants you to analyze images and output whether the image
contains people, animals, scenery, or some other subject.


Statement:

Instagram has launched a new photo recognition service where it applies filters based on what it determines is present in an image uploaded to the service. To do this, it needs to look at pixels and make predictions based on boundaries, colors, and other factors from data in a large set of pre-existing images. This should be a supervised classification problem, where the goal is to determine what category, if any, the image falls into.
