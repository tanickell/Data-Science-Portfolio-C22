## Example from stats lecture

```
samples = [85, 90, 92, 73, 95, 60, 89, 78, 99, 90]
samples
```

Sum ( add up )

```
summation = 0
for value in samples:
  summation += value

summation
```

```
sum(samples)
```

Count ( len of data set )

```
len(samples)
```

```
count = len
count(samples)
```

Mean ( average )

```
summation/len(samples)
```

```
summation/count(samples)
```

Median ( mid-point )

```
sorted(samples)
```

```
sum(sorted(samples)[4:6])/2
```
