## Command structure



Commands tend to have a consistent format:

`command sub-command [options] [arguments]`

The square braces, `[]`, mean that those parts are optional.

Options are usually of the form `-o` or `--option`, that is, a single dash followed by a single letter or a double dash followed by a long name.  Some options can allow a value, e.g. `-o value` or `--option=value`.
