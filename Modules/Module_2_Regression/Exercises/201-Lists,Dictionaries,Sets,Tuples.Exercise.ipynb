## Advanced (Optional- for fun)

You have two lists.  Create a new list with the elements with odd indices from the first list and the elements with even indices from the second list.

```python
list1 = [34, 56, 49, 4900, 249, 494]
list2 = [45, 65, 34, 23, 67, 5]
```


````
list1 = [34, 56, 49, 4900, 249, 494]
list2 = [45, 65, 34, 23, 67, 5]

new_list = list1[0::2] + list2[1::2]
new_list
````

````
# Alternating, using zip and nested list comprehension (for practice):
list1 = [34, 56, 49, 4900, 249, 494]
list2 = [45, 65, 34, 23, 67, 5]

new_list = [i for pair in list(zip(list1[0::2], list2[1::2])) for i in pair]
new_list
````

Starting with a given list, count the occurrences of each unique element.  Create a dictionary to store the count of each element.

```python
list1 = [34, 544, 544, 34, 67, 67, 34, 5, 7, 9]
```


````
my_dict = {}
list1 = [34, 544, 544, 34, 67, 67, 34, 5, 7, 9]
for item in list1:
  my_dict[item] = my_dict.get(item, 0) + 1
my_dict
````

Given a list, check if each element already exists in a dictionary as a key's value.  If not, delete the element from the list.

```python
list1 = [4, 5, 6, 7, 3, 5, 4, 3]
fruit_counts = {
    "apple" : 4,
    "orange": 8,
    "grapefruit": 3,
    "grapes": 7
}
```


````
list1 = [4, 5, 6, 7, 3, 5, 4, 3]
fruit_counts = {
    "apple" : 4,
    "orange": 8,
    "grapefruit": 3,
    "grapes": 7
}

list2 = list1.copy()
for i in list1:
  print(i)
  if i not in list(fruit_counts.values()):
    print(f"{i} not in {fruit_counts.values()}")
    list2.remove(i)
    print(list2)

list1 = list2.copy()
list1
````

With a given dictionary, get all values and add to a list, but don't add duplicates.

```python
fruit = {
  "apple" : 4,
  "orange": 8,
  "grapefruit": 3,
  "grapes": 7,
  "cherries": 4,
  "nanas": 3
}
```


````
fruit = {
  "apple" : 4,
  "orange": 8,
  "grapefruit": 3,
  "grapes": 7,
  "cherries": 4,
  "nanas": 3
}

fruit
````

````
my_set = set()
my_list = []
for i in fruit.values():
  if i not in my_set:
    my_set.add(i)
    my_list.append(i)

my_list
````
